# Offline analysis example
Synthetic time and voltage samples.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

samples = pd.DataFrame({"seconds": [0, 1, 2], "volts": [1.0, 2.0, 3.0]})
assert np.isclose(samples.volts.mean(), 2.0)
samples.plot(x="seconds", y="volts")
plt.savefig("notebook-chart.png")
```
